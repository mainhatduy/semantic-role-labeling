```python
import os
from dotenv import load_dotenv
from huggingface_hub import snapshot_download

# 1. Nạp các biến môi trường từ file .env vào hệ thống
load_dotenv()

# 2. Lấy giá trị token bằng os.getenv
hf_token = os.getenv("HF_TOKEN")

# Kiểm tra xem token đã được load thành công chưa
if not hf_token:
    raise ValueError("Không tìm thấy HF_TOKEN trong file .env hoặc biến môi trường!")

# 3. Cấu hình thông tin repo và thư mục lưu trữ
repo_id = "kiil-lab/ontonotes-releases-5.0"
local_dir = "./my_ontonotes_folder"

print("Đang tiến hành tải dataset từ Hugging Face...")

# 4. Tải repo về local folder sử dụng token vừa lấy được
snapshot_download(
    repo_id=repo_id,
    repo_type="dataset",
    local_dir=local_dir,
    token=hf_token,
    local_dir_use_symlinks=False  # Tải trực tiếp file vật lý về thư mục local_dir
)

print(f" Tải thành công dữ liệu về thư mục: {local_dir}")
```
